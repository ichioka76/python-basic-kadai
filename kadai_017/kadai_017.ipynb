```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            return f"{self.name} is an adult."
        else:
            return f"{self.name} is not an adult."

person1 = Human("takeshi",28)
person2 = Human("takumi",16)
person3 = Human("manabu",22)
person4 = Human("rin",14)

people_list = [person1,person2,person3,person4]

for person in people_list:
    print(person.check_adult())

```
